```python
import pandas as pd

# Đọc file CSV
file_path = r"D:\1. KÌ 1 NĂM 3\1. KHO VÀ KHAI PHÁ DỮ LIỆU\4. CUỐI KÌ\1. DATA\diabetes_binary_5050split_health_indicators_BRFSS2015 - goc.csv" 
data_subset = pd.read_csv(file_path)

# Hiển thị 5 dòng đầu tiên của dữ liệu để kiểm tra
print(data_subset.head())
```

```python
# Giữ lại 10,000 dòng từ trên xuống
top_10000 = data_subset.head(10000)

# Giữ lại 10,000 dòng từ dưới lên
bottom_10000 = data_subset.tail(10000)

# Kết hợp hai phần
data_subset = pd.concat([top_10000, bottom_10000])

# Reset lại chỉ số hàng để dễ dàng xử lý
data_subset.reset_index(drop=True, inplace=True)
```

```python
# Đầu tiên, tạo các cột BMI_1, BMI_2, BMI_3, BMI_4 và gán giá trị 0
data_subset['BMI_1'] = 0
data_subset['BMI_2'] = 0
data_subset['BMI_3'] = 0
data_subset['BMI_4'] = 0

# Áp dụng điều kiện để gán giá trị cho các cột
data_subset.loc[data_subset['BMI'] < 18.5, 'BMI_1'] = 1
data_subset.loc[(data_subset['BMI'] >= 18.5) & (data_subset['BMI'] <= 24.9), 'BMI_2'] = 1
data_subset.loc[(data_subset['BMI'] >= 25) & (data_subset['BMI'] <= 29.9), 'BMI_3'] = 1
data_subset.loc[data_subset['BMI'] >= 30, 'BMI_4'] = 1

# Kiểm tra lại kết quả
print(data_subset[['BMI', 'BMI_1', 'BMI_2', 'BMI_3', 'BMI_4']].head())
```

```python
# Tạo các cột GenHlth_1, GenHlth_2, GenHlth_3, GenHlth_4, GenHlth_5 và gán giá trị mặc định là 0
data_subset['GenHlth_1'] = 0
data_subset['GenHlth_2'] = 0
data_subset['GenHlth_3'] = 0
data_subset['GenHlth_4'] = 0
data_subset['GenHlth_5'] = 0

# Áp dụng điều kiện để gán giá trị cho các cột GenHlth
data_subset.loc[data_subset['GenHlth'] == 1, 'GenHlth_1'] = 1
data_subset.loc[data_subset['GenHlth'] == 2, 'GenHlth_2'] = 1
data_subset.loc[data_subset['GenHlth'] == 3, 'GenHlth_3'] = 1
data_subset.loc[data_subset['GenHlth'] == 4, 'GenHlth_4'] = 1
data_subset.loc[data_subset['GenHlth'] == 5, 'GenHlth_5'] = 1

# Kiểm tra lại kết quả
print(data_subset[['GenHlth', 'GenHlth_1', 'GenHlth_2', 'GenHlth_3', 'GenHlth_4', 'GenHlth_5']].head())
```

```python
# Tạo các cột Sex_1 và Sex_2 và gán giá trị mặc định là 0
data_subset['Sex_1'] = 0 # nữ
data_subset['Sex_2'] = 0 # nam

# Áp dụng điều kiện để gán giá trị cho các cột SexSex
data_subset.loc[data_subset['Sex'] == 0, 'Sex_1'] = 1
data_subset.loc[data_subset['Sex'] == 1, 'Sex_2'] = 1

# Kiểm tra lại kết quả
print(data_subset[['Sex_1', 'Sex_2']].head())
```

```python
# Tạo các cột Age_1 đến Age_7 và gán giá trị mặc định là 0
data_subset['Age_1'] = 0
data_subset['Age_2'] = 0
data_subset['Age_3'] = 0
data_subset['Age_4'] = 0
data_subset['Age_5'] = 0
data_subset['Age_6'] = 0
data_subset['Age_7'] = 0

# Áp dụng điều kiện để gán giá trị cho các cột Age
data_subset.loc[data_subset['Age'].isin([1, 2]), 'Age_1'] = 1
data_subset.loc[data_subset['Age'].isin([3, 4]), 'Age_2'] = 1
data_subset.loc[data_subset['Age'].isin([5, 6]), 'Age_3'] = 1
data_subset.loc[data_subset['Age'].isin([7, 8]), 'Age_4'] = 1
data_subset.loc[data_subset['Age'].isin([9, 10]), 'Age_5'] = 1
data_subset.loc[data_subset['Age'].isin([11, 12]), 'Age_6'] = 1
data_subset.loc[data_subset['Age'] == 13, 'Age_7'] = 1

# Kiểm tra lại kết quả
print(data_subset[['Age', 'Age_1', 'Age_2', 'Age_3', 'Age_4', 'Age_5', 'Age_6', 'Age_7']].head())
```

```python
# Tạo các cột Education_1 đến Education_6 và gán giá trị mặc định là 0
data_subset['Education_1'] = 0
data_subset['Education_2'] = 0
data_subset['Education_3'] = 0
data_subset['Education_4'] = 0
data_subset['Education_5'] = 0
data_subset['Education_6'] = 0

# Áp dụng điều kiện để gán giá trị cho các cột Education
data_subset.loc[data_subset['Education'] == 1, 'Education_1'] = 1
data_subset.loc[data_subset['Education'] == 2, 'Education_2'] = 1
data_subset.loc[data_subset['Education'] == 3, 'Education_3'] = 1
data_subset.loc[data_subset['Education'] == 4, 'Education_4'] = 1
data_subset.loc[data_subset['Education'] == 5, 'Education_5'] = 1
data_subset.loc[data_subset['Education'] == 6, 'Education_6'] = 1

# Kiểm tra lại kết quả
print(data_subset[['Education', 'Education_1', 'Education_2', 'Education_3', 'Education_4', 'Education_5', 'Education_6']].head())
```

```python
# Tạo các cột Income_1 đến Income_8 và gán giá trị mặc định là 0
for i in range(1, 9):
    data_subset[f'Income_{i}'] = 0

# Áp dụng điều kiện để gán giá trị cho các cột Income
for i in range(1, 9):
    data_subset.loc[data_subset['Income'] == i, f'Income_{i}'] = 1

# Kiểm tra lại kết quả
print(data_subset[['Income'] + [f'Income_{i}' for i in range(1, 9)]].head())
```

```python
# Xóa các cột BMI, GenHlth, Age, Education và Income
data_subset = data_subset.drop(columns=['BMI', 'GenHlth', 'Age', 'Education', 'Income', 'Sex', 'PhysHlth', 'MentHlth'])

# Kiểm tra lại DataFrame sau khi xóa
print(data_subset.head())
```

```python
# Đường dẫn lưu file
output_path = r"D:\1. KÌ 1 NĂM 3\1. KHO VÀ KHAI PHÁ DỮ LIỆU\4. CUỐI KÌ\1. DATA\1. DATA CHẠY THUẬT TOÁN\data_done_khac.csv"

# Lưu dữ liệu vào file CSV
data_subset.to_csv(output_path, index=False)

print(f"Dữ liệu đã được lưu tại: {output_path}")
```
