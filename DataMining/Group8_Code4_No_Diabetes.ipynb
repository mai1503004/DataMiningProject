```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

file_path = r"D:\1. KÌ 1 NĂM 3\1. KHO VÀ KHAI PHÁ DỮ LIỆU\4. CUỐI KÌ\1. DATA\1. DATA CHẠY THUẬT TOÁN\data_done_khac.csv"
data = pd.read_csv(file_path)

print("5 dòng đầu của dữ liệu:")
print(data.head())
```

```python
data['Diabetes_binary'] = data['Diabetes_binary'].apply(lambda x: 1 if x == 0 else 0)

print("5 dòng đầu của dữ liệu:")
print(data.head())
```

```python
frequent_itemsets = apriori(data, min_support=0.1, use_colnames=True)

print("Frequent Itemsets:")
print(frequent_itemsets)
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7, num_itemsets=None)

print("Association Rules:")
print(rules)
```

```python
filtered_rules = rules[rules['consequents'] == {'Diabetes_binary'}]

print("Các luật có consequent là 'Diabetes_binary':")
print(filtered_rules)
```

```python
filtered_rules.to_csv(r"D:\1. KÌ 1 NĂM 3\1. KHO VÀ KHAI PHÁ DỮ LIỆU\4. CUỐI KÌ\1. DATA\2. DATA KẾT QUẢ\KETQUA_NO_DIABETES\rules_no_diabetes.csv", index=False)
```
